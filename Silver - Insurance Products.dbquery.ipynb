```sql
CREATE OR REPLACE TABLE insurance.silver.insurance_products AS
SELECT
    id,
    TRIM(product_name) AS product_name,
    TRIM(product_type) AS product_type,
    TRIM(description) AS description,
    CAST(base_premium AS DECIMAL(12,2)) AS base_premium,
    CAST(coverage_amount AS DECIMAL(12,2)) AS coverage_amount,
    created_at
FROM insurance.bronze.insurance_products
WHERE id IS NOT NULL;

SELECT *
FROM insurance.silver.insurance_products;
```
