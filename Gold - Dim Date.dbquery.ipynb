```sql
CREATE OR REPLACE TABLE insurance.gold.dim_date AS
WITH dates AS (
    SELECT EXPLODE(
        SEQUENCE(
            DATE('2025-01-01'),
            DATE('2026-12-31'),
            INTERVAL 1 DAY
        )
    ) AS date
)
SELECT
    date AS date_key,
    YEAR(date) AS year,
    QUARTER(date) AS quarter,
    MONTH(date) AS month,
    MONTHNAME(date) AS month_name,
    DAY(date) AS day,
    DAYOFWEEK(date) AS day_of_week,
    DAYNAME(date) AS day_name
FROM dates;
```
