```sql
CREATE OR REPLACE TABLE insurance.silver.claims AS
SELECT
    id,
    policy_id,
    TRIM(claim_number) AS claim_number,
    claim_date,
    CAST(claim_amount AS DECIMAL(12,2)) AS claim_amount,
    CAST(approved_amount AS DECIMAL(12,2)) AS approved_amount,
    UPPER(TRIM(claim_status)) AS claim_status,
    TRIM(description) AS description,
    created_at
FROM insurance.bronze.claims
WHERE id IS NOT NULL;
```
