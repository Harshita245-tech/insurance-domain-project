```sql
CREATE OR REPLACE TABLE insurance.gold.fact_payment AS
SELECT
    p.id AS payment_key,
    p.policy_id AS policy_key,
    p.payment_date AS payment_date_key,
    p.amount,
    p.payment_method,
    p.payment_status,
    p.transaction_reference,
    p.created_at
FROM insurance.silver.payments p;
```
