```sql
CREATE OR REPLACE TABLE insurance.gold.fact_policy AS
SELECT
    p.id AS policy_key,
    p.customer_id AS customer_key,
    p.product_id AS product_key,
    p.agent_id AS agent_key,
    p.start_date AS start_date_key,
    p.end_date AS end_date_key,
    p.premium_amount,
    p.coverage_amount,
    p.policy_status,
    p.created_at
FROM insurance.silver.policies p;
```
