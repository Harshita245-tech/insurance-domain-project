```sql
USE CATALOG insurance;

CREATE TABLE IF NOT EXISTS insurance.gold.dim_policy AS
SELECT
    id AS policy_key,
    customer_id,
    product_id,
    agent_id,
    policy_number,
    start_date,
    end_date,
    premium_amount,
    coverage_amount,
    policy_status,
    created_at
FROM insurance.silver.policies;
```
