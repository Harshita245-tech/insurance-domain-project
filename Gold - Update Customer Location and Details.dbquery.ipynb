```sql
MERGE INTO insurance.gold.dim_customer AS target
USING insurance.silver.customers AS source
ON target.customer_key = source.id

WHEN MATCHED THEN UPDATE SET
    target.city = source.city,
    target.state = source.state,
    target.first_name = source.first_name,
    target.last_name = source.last_name,
    target.email = source.email,
    target.phone = source.phone,
    target.date_of_birth = source.date_of_birth,
    target.gender = source.gender,
    target.address = source.address,
    target.postal_code = source.postal_code,
    target.created_at = source.created_at

WHEN NOT MATCHED THEN INSERT (
    customer_key, first_name, last_name, email, phone,
    date_of_birth, gender, address, city, state,
    postal_code, created_at
)
VALUES (
    source.id, source.first_name, source.last_name,
    source.email, source.phone, source.date_of_birth,
    source.gender, source.address, source.city,
    source.state, source.postal_code, source.created_at
);
```
