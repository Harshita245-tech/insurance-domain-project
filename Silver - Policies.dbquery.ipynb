```sql
CREATE OR REPLACE TABLE insurance.silver.policies AS
SELECT
    id,
    customer_id,
    product_id,
    agent_id,
    TRIM(policy_number) AS policy_number,
    start_date,
    end_date,
    CAST(premium_amount AS DECIMAL(12,2)) AS premium_amount,
    CAST(coverage_amount AS DECIMAL(12,2)) AS coverage_amount,
    UPPER(TRIM(policy_status)) AS policy_status,
    created_at
FROM insurance.bronze.policies
WHERE id IS NOT NULL;
```
